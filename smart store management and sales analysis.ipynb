# Smart Store Management and Sales Analysis
Retail store want to develop a smart store management and sales analytic system using python. The store mantain information about its products including product id, product name, category, price and available stock. The system should allow the store manager to manage product process customer purchase mantain sales record and analysis sales performance.

Display :

1. View all product
2. Search product
3. Purchase product
4. View sales history
5. Sales analysis
6. Low stock product
7. Exit

# Data of store

```python
low_stock_limit = 10
products = [
    {"product_id": "P001", "name": "Wireless Mouse",       "category": "Electronics", "price": 599.0,  "stock": 45},
    {"product_id": "P002", "name": "Bluetooth Headphones", "category": "Electronics", "price": 1499.0, "stock": 8},
    {"product_id": "P003", "name": "Notebook A5",          "category": "Stationery",  "price": 60.0,   "stock": 120},
    {"product_id": "P004", "name": "Ballpoint Pen Pack",   "category": "Stationery",  "price": 45.0,   "stock": 5},
    {"product_id": "P005", "name": "Ceramic Mug",          "category": "Kitchen",     "price": 249.0,  "stock": 30},
    {"product_id": "P006", "name": "LED Desk Lamp",        "category": "Home",        "price": 899.0,  "stock": 15},
    {"product_id": "P007", "name": "Yoga Mat",             "category": "Fitness",     "price": 799.0,  "stock": 4},
]

sales = []               # each sale is a dictionary
sale_counter = 0          # used to generate simple sale IDs
transaction_counter = 0   # simple stand-in for a timestamp/order number
```

# Helper functions

```python
def find_product(product_id):
    """Return the product dict matching the given ID, or None."""
    for p in products:
        if p["product_id"].lower() == product_id.lower():
            return p
    return None

def format_money(amount):
    return "Rs. " + format(amount,",.2f")

def print_table(rows,headers,widths):
    line=""
    for h,w in zip(headers,widths):
        line +=str(h).ljust(w) + " "
    print(line)
    print("-" * (sum(widths) + 2 * len(widths)))
    for row in rows:
        line=""
        for value,w in zip(row,widths):
            line +=str(value).ljust(w) + " "
        print(line)
```

# View all product

```python
def view_all_products():
    print("\n")
    print(("=") * 30," Products ",("=") * 29)
    print("\n")
    if not products:
        print("No products available.")
        return

    rows=[]
    for p in products:
        stock_dis=str(p["stock"])
        if p["stock"]<=low_stock_limit:
            stock_dis+=" (Low)"
        rows.append([p["product_id"],p["name"],p["category"],format_money(p["price"]),stock_dis])
    print_table(rows,["IDs","Name","Category","Price","Stock"],[6,20,14,17,4])
```

# Search Product

```python
def search_product():
    print("\n")
    print("=" * 27," Search Product","=" * 27)
    print("\n")
    print("Search by: 1.ID 2.Name 3.Category")
    choice=input("Enter your choice 1-3:").strip()

    results=[]
    if choice=="1":
        pid=input("Enter product i'd:").strip()
        p=find_product(pid)
        if p:
            results=[p]

    elif choice=="2":
        pid=input("Enter product name:").strip().lower()
        for p in products:
            if pid in p["name"].lower():
                results.append(p)

    elif choice=="3":
        pid=input("Enter product category:").strip().lower()
        for p in products:
            if pid in p["category"].lower():
                results.append(p)

    else:
        print("Enter valid ID, Name and Category")

    if not results:
        print("No matching products found.")

    print("\n")
    rows=[[p["product_id"],p["name"],p["category"],format_money(p["price"]),p["stock"]]for p in results]
    print_table(rows,["IDs","Name","Category","Price","Stock"],[6,20,14,17,4])
```

# Purchase product

```python
def purchase_product():
    global sale_counter, transaction_counter
 
    print("\n")
    print("=" *23," Purchase Product ","=" *23)
    print("\n")
    pid = input("Enter Product ID: ").strip()
    product=find_product(pid)
 
    if not product:
        print("Product not found.")
        return
        
    print("Selected: " + product["name"] + " | Price: " + format_money(product["price"]) +
          " | Available stock: " + str(product["stock"]))

    if product["stock"] <= 0:
        print("This product is out of stock now.")

    qty_text=input("Enter the quantity to purchase: ").strip()
    if not qty_text.isdigit():
        print("Invalid quantity.")
        return
        
    qty=int(qty_text)
    if qty <= 0:
        print("Enter quantity must be greater than zero.")
        return

    if qty > product["stock"]:
        print("Insufficient stock. only " + str(product["stock"]) + "unit(s) available.")
        return
        
    customer_name=input("Enter your name: ").strip()
    if customer_name == "":
        customer_name="Walk-in Customer"

    total_price=round(qty * product["price"],2)
    product["stock"] -= qty

    sale_counter += 1
    transaction_counter +=1
    sale_id = "S" + str(sale_counter).zfill(4)

    sale_record={
        "sale_id" : sale_id,
        "product_id" : product["product_id"],
        "product_name" : product["name"],
        "category" : product["category"],
        "quantity" : qty,
        "unit_price" : product["price"],
        "total_price" : total_price,
        "customer_name" : customer_name,
        "order_number" : transaction_counter
    }
    sales.append(sale_record)

    print("\nPurchase Successfull")
    print("Sale ID: " + sale_id)
    print(str(qty) + " x " + product["name"] + " = " + format_money(total_price))
    print("Remaining stock: " + str(product["stock"]))

    if product["stock"] <=low_stock_limit:
        print("Note: Stock for '" + product["name"] + "'is now low (" + 
              str(product["stock"]) + " left).")
```

# View Sales History

```python
def view_sale():
    print("\n")
    print("=" * 33," Sales History ","=" * 33)
    print("\n")
    if not sales:
        print("No sales recorded yet.")
        return

    rows = [[s["sale_id"],s["product_name"],s["quantity"],format_money(s["total_price"]),
             s["customer_name"],s["order_number"]] for s in sales]
    print_table(rows,["Sale ID","Product","QTY","Total","Customer","Order"],[9,22,5,15,18,2])

    total_revenue=0.0
    for s in sales:
        total_revenue+=s["total_price"]

    print("\nTotal transactions: " + str(len(sales)))
    print("Total revenue: " + format_money(total_revenue))
```

# Sales Analysis

```python
def sales_analysis():
    print("\n================= Sales Analysis ================")
    print("\n")
    if not sales:
        print("No sales data available for anaylsis.")
        return

    total_revenue=0.0
    total_units=0
    for s in sales:
        total_revenue +=s["total_price"]
        total_units +=s["quantity"]

    total_transactions=len(sales)
    avg_sale_value=total_revenue/total_transactions

    print("Total Revenue: ",format_money(total_revenue))
    print("Total Units sold: ",str(total_units))
    print("Total Transactions: ",str(total_transactions))
    print("Average sale value: ",format_money(avg_sale_value))

    #Best-selling products (by units sold)
    product_qty={}
    product_revenue={}
    for s in sales:
        name=s["product_name"]
        product_qty[name]=product_qty.get(name,0) + s["quantity"]
        product_revenue[name]=product_revenue.get(name,0.0) + s["total_price"]

    print("\n----- Best selling products (by units sold) -----")
    product_names_sorted=sorted(product_qty.keys(), key=lambda n:product_qty[n], reverse=True)

    rows=[]
    for name in product_names_sorted[:5]:
        rows.append([name,product_qty[name],format_money(product_revenue[name])])
    print_table(rows,["Name","Units sold","Revenue"],[20,12,11])

    #Revenue by category
    category_revenue={}
    for s in sales:
        cat=s["category"]
        category_revenue[cat]=category_revenue.get(cat,0.0) + s["total_price"]

    print("\n----- Revenue by category -----")
    categories_sorted=sorted(category_revenue.keys(),key=lambda c:category_revenue[c],reverse=True)
    rows=[[cat, format_money(category_revenue[cat])]for cat in categories_sorted]
    print_table(rows,["Category","Revenue"],[14,13])

    #Top customers by spend
    customer_spend={}
    for s in sales:
        name=s["customer_name"]
        customer_spend[name]=customer_spend.get(name,0.0) + s["total_price"]

    print("\n------- Top customers (by spend) -------")
    customer_sorted=sorted(customer_spend.keys(),key=lambda c:customer_spend[c],reverse=True)
    
    rows=[]
    for cust in customer_sorted:
        rows.append([cust,format_money(customer_spend[cust])])
    print_table(rows,["Name","Total spend"],[22,14])
```

# Low stock products

```python
def low_stock():
    print("\n======== Low Stock Products (threshold <= ",str(low_stock_limit),") =======")
    print("\n")
    low_stock=[p for p in products if p["stock"] <= low_stock_limit]

    if not low_stock:
        print("All products are sufficiently stocked.")

    low_stock_sorted=sorted(low_stock,key=lambda p:p["stock"])
    rows=[[p["product_id"],p["name"],p["category"],p["stock"]]for p in low_stock_sorted]
    print_table(rows,["ID","Name","Category","Stock"],[6,23,14,4])
    print("\n",str(len(low_stock))," products need restocking.")
```

# Main menu

```python
def main():
    menu="""
=================================================
     SMART STORE MANAGEMENT & SALES ANALYSIS
=================================================
1. View All Products
2. Search Product
3. Purchase Product
4. View Sales History
5. Sales Analysis
6. Low Stock Products
7. Exit
=================================================
"""
    while True:
        print(menu)
        choice=input("Enter your choice (1-7): ").strip()

        if choice == "1":
            view_all_products()
        elif choice == "2":
            search_product()
        elif choice == "3":
            purchase_product()
        elif choice == "4":
            view_sale()
        elif choice == "5":
            sales_analysis()
        elif choice == "6":
            low_stock()
        elif choice == "7":
            print("Thank you for using Smart Store Managment System. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 7.")

        input("\nPress Enter to continue...")
        
if __name__ == "__main__":
    main()
```
